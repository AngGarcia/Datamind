```python
# %pip install opencv-python # El uso de %pip en lugar de !pip garantiza que se instale exactamente en el kernel que tienes abierto.
# hemos usado tf_env porque es python versión 3.10
# No olvidar activar el entorno para instalar cositas. EN ANACONDA PROMPT (en esta terminal da error porque las cosas de conda no están en el path)
# conda activate tf_env

import cv2 # conda install opencv-python
import os

IMAGE_PATH = "../../ImagesData/Data_cartas_multiples/images/images/train"
LABEL_PATH = "../../ImagesData/Data_cartas_multiples/labels/train"

# cargar clases
with open("classes.txt") as f:
    classes = [line.strip() for line in f.readlines()]

count = 0
max_images = 10

for img_file in os.listdir(IMAGE_PATH):

    if count >= max_images:
        break

    if not img_file.endswith(".jpg"):
        continue

    img_path = os.path.join(IMAGE_PATH, img_file)
    label_path = os.path.join(LABEL_PATH, img_file.replace(".jpg", ".txt"))

    # leer imagen
    img = cv2.imread(img_path)

    if img is None:
        continue

    # ⚠️ IMPORTANTE: copiamos la imagen (no tocamos la original)
    img_copy = img.copy()

    h, w, _ = img_copy.shape

    if os.path.exists(label_path):
        with open(label_path) as f:
            lines = f.readlines()

        for line in lines:
            cls, x, y, bw, bh = map(float, line.split())

            # convertir a píxeles
            x1 = int((x - bw/2) * w)
            y1 = int((y - bh/2) * h)
            x2 = int((x + bw/2) * w)
            y2 = int((y + bh/2) * h)

            cls = int(cls)

            # dibujar caja
            cv2.rectangle(img_copy, (x1, y1), (x2, y2), (0,255,0), 2)

            # texto clase
            label = classes[cls]
            cv2.putText(img_copy, label, (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 2)

    # mostrar imagen
    cv2.imshow("Validación", img_copy)

    key = cv2.waitKey(0)

    # salir si pulsas ESC
    if key == 27:
        break

    count += 1

cv2.destroyAllWindows()
```
