```python
import ast

def sacar_columnas(df):

  # Convertir los strings a listas reales
  if isinstance(df["player_hand"].iloc[0], str):
      manos_listas = df["player_hand"].apply(ast.literal_eval)
  else:
      manos_listas = df["player_hand"]

  # Hacer un df con la serie de listas
  df_cartas = pd.DataFrame(manos_listas.tolist(), index=df.index)

  # Renombrar las columnas de 0-7 a carta_1-carta_8
  df_cartas.columns = [f"carta_{i+1}" for i in range(df_cartas.shape[1])]

  # Sustituir NaN por 0
  df_cartas = df_cartas.fillna(0).astype(int)

  # Juntar las 8 columnas y borrar 'player_hand'
  df = pd.concat([df, df_cartas], axis=1).drop(columns=["player_hand"])

  return df
```

```python
import xgboost as xgb

modelo_cargado = xgb.XGBClassifier()

modelo_cargado.load_model("modelo_blackjack.json")
```

```python
def predecir_accion(
    cartas_jugador,
    dealer_up,
    cards_remaining=200,
    run_count=0,
    true_count=0,
    player_final_action_value=None,
):
    """Recibe tus cartas y te dice la acción óptima según el modelo."""

    # Si no se especifica la suma, la calculamos nosotros sumando la lista
    if player_final_action_value is None:
        player_final_action_value = sum(cartas_jugador)

    # Lista de 8 posiciones rellena de ceros (teníamos 8 cartas)
    vector_cartas = [0] * 8

    # Metemos las cartas en las primeras posiciones
    for i, carta in enumerate(cartas_jugador[:8]):
        vector_cartas[i] = carta

    # Diccionario con todas las columnas que teníamos en el entrenamiento
    datos_jugada = {
        "cards_remaining": [cards_remaining],
        "dealer_up": [dealer_up],
        "player_final_action_value": [player_final_action_value],
        "run_count": [run_count],
        "true_count": [true_count],
    }

    # Añadimos las 8 columnas de las cartas (carta_1, carta_2...)
    for i in range(8):
        datos_jugada[f"carta_{i+1}"] = [vector_cartas[i]]

    # Hacerlo DataFrame
    df_jugada = pd.DataFrame(datos_jugada)

    # Orden de las columnas
    columnas_ordenadas = [
        "cards_remaining",
        "dealer_up",
        "player_final_action_value",
        "run_count",
        "true_count",
    ] + [f"carta_{i+1}" for i in range(8)]
    df_jugada = df_jugada[columnas_ordenadas]

    # Predecir el número de clase
    prediccion_numerica = modelo_cargado.predict(df_jugada)[0]

    # Traducir el número a la letra ('S', 'H', 'P'...)
    accion_recomendada = encoder.inverse_transform([prediccion_numerica])[0]

    # Ver las probabilidades de cada acción
    probabilidades = modelo_cargado.predict_proba(df_jugada)[0]
    mapeo_probs = dict(zip(encoder.classes_, probabilidades))

    print(f"Mano del jugador: {cartas_jugador} (Suma: {player_final_action_value})")
    print(f"Carta visible del Dealer: {dealer_up}")
    print(f"ACCIÓN RECOMENDADA: {accion_recomendada}")
    print(f"Confianza en la decisión: {mapeo_probs[accion_recomendada]*100:.2f}%")

    return accion_recomendada
```
